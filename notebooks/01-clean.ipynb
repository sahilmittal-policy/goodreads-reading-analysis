```python
import pandas as pd
import numpy as np 

data = pd.read_csv("../data/raw/goodreads_library_export.csv")

data.columns
my_data = data.drop(columns= ['Author l-f', 'ISBN', 'ISBN13', 'Publisher', 
                               'Binding', 'My Review', 'Spoiler', 'Private Notes',  
                               'Owned Copies', 'Year Published', 'Date Added', 
                               'Bookshelves', 'Bookshelves with positions', 'Read Count', 'read_dates', 'Additional Authors'])


read_books = my_data[(my_data['Exclusive Shelf'] == 'read')]
read_books['date_read'] = pd.to_datetime(read_books['Date Read'])
read_books['year_read'] = read_books['date_read'].dt.year
read_books['month_year_read'] = read_books['date_read'].dt.strftime('%B %Y')
read_books = read_books.astype({'Number of Pages': 'Int64', 'Original Publication Year': 'Int64', 'n_ratings': 'Int64'})
read_books = read_books.drop(columns= ['Date Read', 'Exclusive Shelf', 'Book Id'])
split_genres = read_books['genres'].str.split(r'[|;]', expand=True, n=3)
read_books['genre_1'] = split_genres[0]
read_books['genre_2'] = split_genres[2]
read_books = read_books.drop(columns= ['genres'])
read_books.loc[4, 'Author'] = 'Kurt Vonnegut Jr.'
read_books = read_books.sort_values(by = 'date_read').reset_index(drop=True)

read_books.dtypes
```

```python
read_books.to_csv(r'../data/Processed/read_books.csv', index=False)
```
